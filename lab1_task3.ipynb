```python
import random

class VacuumEnvironment:
    """Two-square world: locations 'A' and 'B', each Clean (0) or Dirty (1)."""
    def __init__(self, status=None, location=None):
        self.status = status if status else {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = location if location else random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        """Applies the action and returns the reward for that step."""
        if action == 'Suck':
            was_dirty = self.status[self.agent_location] == 1
            self.status[self.agent_location] = 0
            return 10 if was_dirty else 0   # reward only if something was actually cleaned
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        return -1                            # cost of moving

def reflex_vacuum_agent(percept):
    """Condition-action rules: decide using ONLY the current percept."""
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

def random_vacuum_agent(percept):
    """Ignores the percept and picks any action at random."""
    return random.choice(['Left', 'Right', 'Suck'])

def run_agent(agent, status, location, steps=20):
    env = VacuumEnvironment(dict(status), location)   # copy so each agent gets the same start
    score = 0
    for _ in range(steps):
        action = agent(env.percept())
        score += env.execute(action)
    return score

# ---- Experiment: 10 random initial states, 20 steps each ----
random.seed(7)
STEPS, TRIALS = 20, 10
reflex_scores, random_scores = [], []

print(f"{'Trial':<6}{'Initial state':<30}{'Reflex':>8}{'Random':>8}")
for t in range(1, TRIALS + 1):
    status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
    location = random.choice(['A', 'B'])

    r = run_agent(reflex_vacuum_agent, status, location, STEPS)
    n = run_agent(random_vacuum_agent, status, location, STEPS)
    reflex_scores.append(r)
    random_scores.append(n)

    print(f"{t:<6}{str(status) + ' @' + location:<30}{r:>8}{n:>8}")

print("\nAverage score over", TRIALS, "runs:")
print("  Simple Reflex Agent:", sum(reflex_scores) / TRIALS)
print("  Random Agent       :", sum(random_scores) / TRIALS)
```
